```python
import os
import h5py
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
path2filternet_8dir_trials = "core/filternet_8dir_10trials"

angles = [0, 45, 90, 135, 180, 225, 270, 315]
trials = range(10)
```

```python
for triali in trials:
    all_timestamps = []
    all_node_ids = []

    file_name = f"8dir_trial{triali}_lgn_spikes.h5"

    for ii, angle in enumerate(angles):
        path2file = os.path.join(
            path2filternet_8dir_trials, f"angle{angle}_trial{triali}", "spikes.h5"
        )
        with h5py.File(path2file, "r") as f:
            timestamps = f["/spikes/lgn/timestamps"][:] + (
                3000 * ii
            )  # + 3s = 3000ms for each angle simulated trial
            node_ids = f["/spikes/lgn/node_ids"][:]
            all_timestamps.append(timestamps)
            all_node_ids.append(node_ids)

    with h5py.File(os.path.join(path2filternet_8dir_trials, file_name), "w") as f:
        f.create_dataset("/spikes/lgn/timestamps", data=np.concatenate(all_timestamps))
        f.create_dataset("/spikes/lgn/node_ids", data=np.concatenate(all_node_ids))
```

```python
with h5py.File(os.path.join(path2filternet_8dir_trials, file_name), "r") as f:
    timestamps = f["/spikes/lgn/timestamps"][:]
    node_ids = f["/spikes/lgn/node_ids"][:]

import seaborn as sns

spike_lgn_df = pd.DataFrame(
    {
        "timestamps": timestamps,
        "node_ids": node_ids,
    }
)

fig, ax = plt.subplots(figsize=(10, 5))
sns.scatterplot(
    data=spike_lgn_df,
    x="timestamps",
    y="node_ids",
    hue="node_ids",
    s=1,
    linewidth=0,
    alpha=0.3,
    ax=ax,
)
# ax.set_xlim([0,500])
ax.invert_yaxis()
ax.legend(bbox_to_anchor=(1.02, 1), loc="upper left", borderaxespad=0)
ax.set_xlabel("time [ms]")
```
